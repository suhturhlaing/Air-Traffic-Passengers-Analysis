1. How has passenger traffic changed over time?
  • Columns: Activity Period, Passenger Count, Year, Month
  • Graph: Line Chart
  • Description: Plot passenger count over time to identify trends and seasonal patterns.

```python
# Convert Activity Period to datetime
df['Activity Period'] = pd.to_datetime(df['Activity Period'], format='%Y%m')

# Check the result
df[['Activity Period']].head()
```

```python
# line chart analysis
import plotly.express as px

df['Activity Period'] = pd.to_datetime(df['Activity Period'], format='%Y%m')
df['Activity Period Label'] = df['Activity Period'].dt.strftime('%Y-%b')

monthly_passengers = df.groupby(['Activity Period', 'Activity Period Label'])['Passenger Count'].sum().reset_index()

fig = px.line(
    monthly_passengers,
    x='Activity Period Label',
    y='Passenger Count',
    title='✈️ Passenger Traffic Over Time',
    labels={'Activity Period Label': 'Date', 'Passenger Count': 'Total Passengers'},
    markers=True
)

# Customize layout
fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    xaxis_tickangle=-45,
    )
fig.show()


```

2. Which airlines have the highest passenger counts?
• Columns: Operating Airline, Passenger Count
• Graph: Bar Chart
• Description: Compare the total passenger count for each airline

```python
# Group by airline and sum passenger counts
airline_totals = df.groupby('Operating Airline')['Passenger Count'].sum().reset_index()

# Sort by passenger count in descending order and get top 15 airlines
top_airlines = airline_totals.sort_values('Passenger Count', ascending=False).head(15)

# Create bar chart
fig = px.bar(
    top_airlines,
   x='Passenger Count',
   y='Operating Airline',
    orientation='h',
    title='✈️ Top Airlines by Passenger Count',
    labels={'Passenger Count': 'Total Passengers', 'Operating Airline': 'Airline'},
    color='Passenger Count',
    color_continuous_scale='viridis'
)


fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    yaxis={'categoryorder': 'total ascending'},
    height=600
)


fig.update_xaxes(tickformat=',')
fig.show()
```

3. What is the geographic distribution of passengers?
• Columns: GEO Summary, GEO Region, Passenger Count
• Graph: Pie Chart or Choropleth Map
• Description: Show the proportion of passengers from different geographic regions or a map
highlighting passenger counts by region

```python
# Geographic distribution by GEO Summary (Domestic vs International)
geo_summary = df.groupby('GEO Summary')['Passenger Count'].sum().reset_index()

fig1 = px.pie(
    geo_summary,
    values='Passenger Count',
    names='GEO Summary',
    title='🌍 Passenger Distribution: Domestic vs International',
    color='GEO Summary',
    color_discrete_map={'Domestic': '#1f77b4', 'International': '#ff7f0e'}
)
fig1.update_layout(template='plotly_white', title_x=0.5)
fig1.show()

```

```python
# Geographic distribution by GEO Region
geo_region = df.groupby('GEO Region')['Passenger Count'].sum().reset_index()
geo_region = geo_region.sort_values('Passenger Count', ascending=False)

# Create pie chart for GEO Region
fig2 = px.pie(
    geo_region,
    values='Passenger Count',
    names='GEO Region',
    title='🌍 Passenger Distribution by Geographic Region',
    color='GEO Region',
    hover_data=['Passenger Count'],
    labels={'Passenger Count': 'Passengers'}
)

fig2.update_layout(
    template='plotly_white',
    title_x=0.5,
    height=600
)

# Show values and percentages on the chart
fig2.update_traces(
    textposition='inside',
    textinfo='percent+label',
    insidetextorientation='radial'
)

fig2.show()
```

```python
import plotly.express as px
import pandas as pd

# Filter data for International and Domestic only
filtered_df = df[df['GEO Summary'].isin(['International', 'Domestic'])]

# Group by GEO Region and GEO Summary
geo_choropleth = filtered_df.groupby(['GEO Region', 'GEO Summary'])['Passenger Count'].sum().reset_index()

# Create mapping from GEO Region to country/continent codes
region_mapping = {
    'US': 'USA',
    'Canada': 'CAN',
    'Mexico': 'MEX',
    'Central America': 'CYM',
    'Europe': 'EUU',
    'Asia': 'ASI',
    'Australia / Oceania': 'AUS'
}


geo_choropleth['Country_Code'] = geo_choropleth['GEO Region'].map(region_mapping)

# Create choropleth map
fig = px.choropleth(
    geo_choropleth,
    locations='Country_Code',
    color='Country_Code',
    hover_name='GEO Region',
    hover_data={'GEO Summary': True, 'Passenger Count': True, 'Country_Code': False},
    title='🌍 Passenger Distribution by Geographic Region (Choropleth Map)',
    color_continuous_scale='viridis',
    scope='world'
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    height=600,
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection_type='equirectangular'
    )
)

# Add annotations for regions without standard country codes
fig.add_annotation(
    x=0.5, y=0.02,
    xref='paper', yref='paper',
    text="Note: Asia and Europe shown as aggregated regions",
    showarrow=False,
    bgcolor='white',
    bordercolor='black',
    borderwidth=1
)

fig.show()
```

```python
import plotly.graph_objects as go

filtered_df = df[df['GEO Summary'].isin(['International', 'Domestic'])]

# Create separate dataframes for Domestic and International
domestic_data = filtered_df[filtered_df['GEO Summary'] == 'Domestic'].groupby('GEO Region')['Passenger Count'].sum().reset_index()
international_data = filtered_df[filtered_df['GEO Summary'] == 'International'].groupby('GEO Region')['Passenger Count'].sum().reset_index()

# Sort regions by total passenger count
all_regions = filtered_df.groupby('GEO Region')['Passenger Count'].sum().sort_values(ascending=False).index

# Create the figure
fig = go.Figure()

# Add Domestic bars
fig.add_trace(go.Bar(
    name='Domestic',
    x=domestic_data['GEO Region'],
    y=domestic_data['Passenger Count'],
    marker_color='#1f77b4',
    text=domestic_data['Passenger Count'],
    texttemplate='%{text:,}',
    textposition='outside'
))

# Add International bars
fig.add_trace(go.Bar(
    name='International',
    x=international_data['GEO Region'],
    y=international_data['Passenger Count'],
    marker_color='#ff7f0e',
    text=international_data['Passenger Count'],
    texttemplate='%{text:,}',
    textposition='outside'
))


fig.update_layout(
    title='🌍 Passenger Count by Geographic Region - Domestic vs International',
    xaxis_title='Geographic Region',
    yaxis_title='Total Passengers',
    template='plotly_white',
    title_x=0.5,
    height=500,
    xaxis={'categoryorder': 'array', 'categoryarray': all_regions},
    barmode='group',  # This creates grouped bars
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)


fig.update_yaxes(tickformat=',')

fig.show()
```

•Columns: Terminal, Boarding Area, Passenger Count

```python
# Group by Terminal and Boarding Area
terminal_boarding = df.groupby(['Terminal', 'Boarding Area'])['Passenger Count'].sum().reset_index()

# Create a pivot table for heatmap
pivot_table = terminal_boarding.pivot_table(
    values='Passenger Count',
    index='Terminal',
    columns='Boarding Area',
    fill_value=0
)


# Create heatmap
fig1 = px.imshow(
    pivot_table,
    title='🔥 Passenger Distribution: Terminal vs Boarding Area (Heatmap)',
    labels=dict(x="Boarding Area", y="Terminal", color="Passengers"),
    #color_continuous_scale=colorscale,
    #color_continuous_scale = 'viridis',
    color_continuous_scale='reds',
    aspect="auto"
)

fig1.update_layout(template='plotly_white', title_x=0.5)
fig1.show()
```

7. Which months have the highest passenger traffic?
•Columns: Month, Passenger Count
•Graph: Bar Chart
•Description: Show passenger counts for each month to identify peak travel periods.

```python
# Group by Month and sum passenger counts
monthly_totals = df.groupby('Month')['Passenger Count'].sum().reset_index()

# Define the correct month order for sorting
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
              'July', 'August', 'September', 'October', 'November', 'December']

# Convert Month to categorical with proper ordering
monthly_totals['Month'] = pd.Categorical(monthly_totals['Month'], categories=month_order, ordered=True)
monthly_totals = monthly_totals.sort_values('Month')

# Create bar chart
fig = px.bar(
    monthly_totals,
    x='Month',
    y='Passenger Count',
    title='📅 Monthly Passenger Traffic',
    labels={'Passenger Count': 'Total Passengers', 'Month': 'Month'},
    color='Passenger Count',
    color_continuous_scale='viridis'
)


fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    xaxis_tickangle=45,
    height=500
)

# Format y-axis to show numbers with commas
fig.update_yaxes(tickformat=',')

# Add value labels on bars
fig.update_traces(
    texttemplate='%{y:,}',
    textposition='outside'
)

fig.show()
```

8. How have adjusted passenger counts changed over time?
•Columns: Activity Period, Adjusted Passenger Count, Year, Month
•Graph: Line Chart
•Description: Plot adjusted passenger counts over time to see trends and adjustments

```python
# Group by Activity Period and sum adjusted passenger counts
adjusted_totals = df.groupby(['Activity Period', 'Activity Period Label'])['Adjusted Passenger Count'].sum().reset_index()

# Create line chart for adjusted passenger counts
fig = px.line(
    adjusted_totals,
    x='Activity Period Label',
    y='Adjusted Passenger Count',
    title='📈 Adjusted Passenger Counts Over Time',
    labels={'Adjusted Passenger Count': 'Adjusted Passengers', 'Activity Period Label': 'Date'},
    markers=True
)


fig.update_layout(
    template='plotly_white',
    title_x=0.5,
   xaxis_tickangle= -60,
    height=500
)

# Format y-axis to show numbers with commas
fig.update_yaxes(tickformat=',')

# Add trendline
fig.add_scatter(
    x=adjusted_totals['Activity Period Label'],
    y=adjusted_totals['Adjusted Passenger Count'].rolling(window=3).mean(),
    mode='lines',
    line=dict(dash='dot', color='red', width=2),
    name='3-Month Moving Average'
)

fig.show()
```
